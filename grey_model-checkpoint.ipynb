```python
import os
import cv2
import numpy as np
import matplotlib.pyplot as plt
from glob import glob
from sklearn.model_selection import train_test_split
```

```python
sar_path = '../v_2/agri/s1/'
rgb_path = '../v_2/agri/s2/'

sar_images = sorted(glob(os.path.join(sar_path, '*.png')))
rgb_images = sorted(glob(os.path.join(rgb_path, '*.png')))

print(f'Total SAR images: {len(sar_images)}')
print(f'Total RGB images: {len(rgb_images)}')
```

```python
#to compare sar and rgb from the dataset side by side 
def load_images(index=0):
    sar = cv2.imread(sar_images[index], cv2.IMREAD_GRAYSCALE)
    rgb = cv2.cvtColor(cv2.imread(rgb_images[index]), cv2.COLOR_BGR2RGB)
    return sar, rgb

sar, rgb = load_images(0)

plt.figure(figsize=(10, 5))
plt.subplot(1, 2, 1)
plt.title('SAR (Grayscale)')
plt.imshow(sar, cmap='gray')
plt.subplot(1, 2, 2)
plt.title('RGB (Target)')
plt.imshow(rgb)
plt.show()
```

```python
IMG_SIZE = 128

def preprocess(index):
    sar = cv2.resize(cv2.imread(sar_images[index], cv2.IMREAD_GRAYSCALE), (IMG_SIZE, IMG_SIZE))
    rgb = cv2.resize(cv2.cvtColor(cv2.imread(rgb_images[index]), cv2.COLOR_BGR2RGB), (IMG_SIZE, IMG_SIZE))
    
    sar = sar.astype('float32') / 255.0
    rgb = rgb.astype('float32') / 255.0

    sar = np.expand_dims(sar, axis=-1)   # (H, W, 1)
    
    return sar, rgb

# Prepare full dataset
X, Y = [], []
for i in range(len(sar_images)):
    sar, rgb = preprocess(i)
    X.append(sar)
    Y.append(rgb)

X = np.array(X)
Y = np.array(Y)

print("Shape of SAR input:", X.shape)
print("Shape of RGB target:", Y.shape)
```

*U-Net Model for Colorization

```python
import torch
import torch.nn as nn
import torch.nn.functional as F

class UNet(nn.Module):
    def __init__(self, in_channels=1, out_channels=3):
        super(UNet, self).__init__()

        # Encoder
        self.enc1 = self.conv_block(in_channels, 64)
        self.pool1 = nn.MaxPool2d(2)

        self.enc2 = self.conv_block(64, 128)
        self.pool2 = nn.MaxPool2d(2)

        self.enc3 = self.conv_block(128, 256)
        self.pool3 = nn.MaxPool2d(2)

        # Bottleneck
        self.bottleneck = self.conv_block(256, 512)

        # Decoder
        self.up3 = nn.Upsample(scale_factor=2, mode='bilinear', align_corners=True)
        self.dec3 = self.conv_block(512 + 256, 256)

        self.up2 = nn.Upsample(scale_factor=2, mode='bilinear', align_corners=True)
        self.dec2 = self.conv_block(256 + 128, 128)

        self.up1 = nn.Upsample(scale_factor=2, mode='bilinear', align_corners=True)
        self.dec1 = self.conv_block(128 + 64, 64)

        # Output layer
        self.final = nn.Conv2d(64, out_channels, kernel_size=1)
        self.output_activation = nn.Sigmoid()  # for pixel values [0, 1]

    def conv_block(self, in_channels, out_channels):
        return nn.Sequential(
            nn.Conv2d(in_channels, out_channels, kernel_size=3, padding=1),
            nn.ReLU(inplace=True),
            nn.Conv2d(out_channels, out_channels, kernel_size=3, padding=1),
            nn.ReLU(inplace=True)
        )

    def forward(self, x):
        # Encoder
        e1 = self.enc1(x)
        p1 = self.pool1(e1)

        e2 = self.enc2(p1)
        p2 = self.pool2(e2)

        e3 = self.enc3(p2)
        p3 = self.pool3(e3)

        # Bottleneck
        b = self.bottleneck(p3)

        # Decoder
        d3 = self.up3(b)
        d3 = torch.cat((d3, e3), dim=1)
        d3 = self.dec3(d3)

        d2 = self.up2(d3)
        d2 = torch.cat((d2, e2), dim=1)
        d2 = self.dec2(d2)

        d1 = self.up1(d2)
        d1 = torch.cat((d1, e1), dim=1)
        d1 = self.dec1(d1)

        out = self.final(d1)
        return self.output_activation(out)

```

```python
from sklearn.model_selection import train_test_split

# Assuming X and Y are your full datasets from preprocessing step
X_train, X_val, Y_train, Y_val = train_test_split(X, Y, test_size=0.2, random_state=42)

print("Training set:", X_train.shape, Y_train.shape)
print("Validation set:", X_val.shape, Y_val.shape)
```

TRAIN THE MODEL

```python
from torch.utils.data import Dataset, DataLoader
import torch

class SARColorizationDataset(Dataset):
    def __init__(self, X, Y):
        self.X = torch.tensor(X, dtype=torch.float32).permute(0, 3, 1, 2)  # [N, C, H, W]
        self.Y = torch.tensor(Y, dtype=torch.float32).permute(0, 3, 1, 2)

    def __len__(self):
        return len(self.X)

    def __getitem__(self, idx):
        return self.X[idx], self.Y[idx]

train_dataset = SARColorizationDataset(X_train, Y_train)
val_dataset = SARColorizationDataset(X_val, Y_val)

train_loader = DataLoader(train_dataset, batch_size=8, shuffle=True)
val_loader = DataLoader(val_dataset, batch_size=8, shuffle=False)
```

```python
import torch.nn as nn
import torch.optim as optim
from tqdm import tqdm

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
model = UNet(in_channels=1, out_channels=3).to(device)

criterion = nn.MSELoss()
optimizer = optim.Adam(model.parameters(), lr=0.001)

num_epochs = 30
best_val_loss = float('inf')

for epoch in range(num_epochs):
    model.train()
    train_loss = 0.0
    train_loop = tqdm(train_loader, desc=f"[Epoch {epoch+1}/{num_epochs}] Training")

    for i, (inputs, targets) in enumerate(train_loop):
        inputs, targets = inputs.to(device), targets.to(device)

        optimizer.zero_grad()
        outputs = model(inputs)
        loss = criterion(outputs, targets)
        loss.backward()
        optimizer.step()

        train_loss += loss.item()
        train_loop.set_postfix(loss=f"{train_loss / (i+1):.4f}")

    # Validation
    model.eval()
    val_loss = 0.0
    with torch.no_grad():
        for inputs, targets in val_loader:
            inputs, targets = inputs.to(device), targets.to(device)
            outputs = model(inputs)
            loss = criterion(outputs, targets)
            val_loss += loss.item()

    avg_train_loss = train_loss / len(train_loader)
    avg_val_loss = val_loss / len(val_loader)

    print(f"Epoch {epoch+1} - Train Loss: {avg_train_loss:.4f} | Val Loss: {avg_val_loss:.4f}")

    # Save the best model
    if avg_val_loss < best_val_loss:
        best_val_loss = avg_val_loss
        torch.save(model.state_dict(), 'best_unet_model.pth')
        print("✅ Saved best model so far!")
```

```python
from skimage.metrics import peak_signal_noise_ratio as psnr
from skimage.metrics import structural_similarity as ssim
import numpy as np
import matplotlib.pyplot as plt
import torch

model.eval()
total_psnr = 0
total_ssim = 0
n = len(val_loader.dataset)

with torch.no_grad():
    for inputs, targets in val_loader:
        inputs = inputs.to(device)
        outputs = model(inputs).cpu().numpy()
        targets = targets.numpy()

        for i in range(outputs.shape[0]):
            pred_img = outputs[i].transpose(1, 2, 0)  # [H, W, C]
            true_img = targets[i].transpose(1, 2, 0)

            # Scale to [0, 255]
            pred_img_uint8 = (pred_img * 255).clip(0, 255).astype(np.uint8)
            true_img_uint8 = (true_img * 255).clip(0, 255).astype(np.uint8)

            # Compute PSNR & SSIM
            total_psnr += psnr(true_img_uint8, pred_img_uint8, data_range=255)
            total_ssim += ssim(true_img_uint8, pred_img_uint8, data_range=255, channel_axis=-1)

# Average results
avg_psnr = total_psnr / n
avg_ssim = total_ssim / n

print(f"✅ Average PSNR: {avg_psnr:.2f} dB")
print(f"✅ Average SSIM: {avg_ssim:.4f}")
```
